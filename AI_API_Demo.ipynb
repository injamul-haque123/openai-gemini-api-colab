# OpenAI + Google Gemini API Demo

This notebook runs in Google Colab and demonstrates how to use OpenAI and Google Gemini APIs securely with Colab Secrets.

## Setup
1. In Colab, open **Secrets** (key icon) and add `OPENAI_API_KEY` and/or `GEMINI_API_KEY`.
2. Run the installation cell.
3. Run the configuration and example cells.

**Never put API keys directly in this notebook or commit them to GitHub.**

```python
from google.colab import userdata

def get_secret(name):
    try:
        return userdata.get(name)
    except Exception:
        return None

OPENAI_API_KEY = get_secret("OPENAI_API_KEY")
GEMINI_API_KEY = get_secret("GEMINI_API_KEY")

print("OpenAI key:", "configured" if OPENAI_API_KEY else "not configured")
print("Gemini key:", "configured" if GEMINI_API_KEY else "not configured")
```

```python
from openai import OpenAI
from google import genai

openai_client = OpenAI(api_key=OPENAI_API_KEY) if OPENAI_API_KEY else None
gemini_client = genai.Client(api_key=GEMINI_API_KEY) if GEMINI_API_KEY else None

def ask_openai(prompt, model="gpt-5.6-luna"):
    if not openai_client:
        raise RuntimeError("OPENAI_API_KEY is not configured in Colab Secrets.")
    response = openai_client.responses.create(
        model=model,
        input=prompt,
    )
    return response.output_text

def ask_gemini(prompt, model="gemini-3.8-flash"):
    if not gemini_client:
        raise RuntimeError("GEMINI_API_KEY is not configured in Colab Secrets.")
    response = gemini_client.models.generate_content(
        model=model,
        contents=prompt,
    )
    return response.text

def ask_ai(provider, prompt):
    provider = provider.lower().strip()
    if provider == "openai":
        return ask_openai(prompt)
    if provider == "gemini":
        return ask_gemini(prompt)
    raise ValueError("Provider must be 'openai' or 'gemini'.")
```

```python
# Example: choose a provider and ask a question
provider = "openai"   # Change to "gemini"
prompt = "Explain machine learning in simple words."

try:
    answer = ask_ai(provider, prompt)
    print(answer)
except Exception as e:
    print(f"Error: {e}")
```

```python
# Optional interactive loop
while True:
    provider = input("Provider (openai/gemini/exit): ").strip().lower()
    if provider == "exit":
        break
    prompt = input("Question: ")
    try:
        print("\n" + ask_ai(provider, prompt) + "\n")
    except Exception as e:
        print(f"Error: {e}")
```
